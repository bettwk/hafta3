```python

filmler = {}

while True:
    print("\n--- FILM FESTIVALI ---")
    print("1 - Film Ekle")
    print("2 - Filmleri Listele")
    print("3 - Puan Ver")
    print("4 - Ortalama Goster")
    print("5 - Kazanan Film")
    print("6 - Film Sil")
    print("0 - Cikis")

    secim = input("Seciminiz: ")

    if secim == "1":
        film_adi = input("Film adi: ")

        if film_adi in filmler:
            print("Bu film zaten ekli!")
        else:
            filmler[film_adi] = []
            print("Film eklendi.")

    elif secim == "2":
        for film_adi in filmler:
            print(film_adi)
        pass

    elif secim == "3":
        film_adi = input("puan girmek istediginiz filmin adini yaziniz: ")
        if film_adi not in filmler:
            input("bu film listede yok, lutfen baska bir film secin: ")
        else:
            puan = int(input("1-10 arasi bir puan giriniz:"))
            filmler[film_adi].append(puan)        
        pass

    elif secim == "4":
        film_adi= input("ortalamasini gormek istedigin filmin adini yaz: ")
        toplam = 0
        if film_adi in filmler:
            for puan in filmler[film_adi]:
                toplam += int(puan)
        ortalama = toplam/len(filmler[film_adi])
        print(f"{film_adi} filminin ortalama puani : {ortalama}")
        pass

    elif secim == "5":
        en_yuksek_ortalama = -1
        en_iyi_film = ""

        for film_adi in filmler:
            toplam = 0
            puanlar = filmler[film_adi]

            if len(puanlar) == 0:
                continue

            for puan in puanlar:
                toplam += puan

            ortalama = toplam / len(puanlar)

            if ortalama > en_yuksek_ortalama:
                en_yuksek_ortalama = ortalama
                en_iyi_film = film_adi

        if en_iyi_film != "":
            print(f"En yüksek ortalamalı film: {en_iyi_film}")
            print(f"Ortalaması: {en_yuksek_ortalama:.2f}")
        else:
            print("Henüz puan verilmiş bir film yok.")

    elif secim == "6":
        film_adi = input("silmek istediginiz film adini yazin:")
        if film_adi in filmler:
            del filmler[film_adi]
        pass

    elif secim == "0":
        print("Festival kapatildi.")
        break

    else:
        print("Gecersiz secim!")
```
