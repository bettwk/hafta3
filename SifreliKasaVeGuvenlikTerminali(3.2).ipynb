```python

import random

kasalar = {}
puan = 0

while True:
    print("\n--- GUVENLIK TERMINALI ---")
    print("1 - Yeni Kasa Olustur")
    print("2 - Kasa Ac")
    print("3 - Kasalari Listele")
    print("4 - Puanimi Goster")
    print("5 - Kilitli Kasalar")
    print("0 - Cikis")

    secim = input("Secim: ")

    if secim == "1":
        kasa_adi = input("Kasa adi: ")

        if kasa_adi in kasalar:
            print("Bu kasa zaten var!")
            continue

        sifre = random.randint(1000, 9999)

        kasalar[kasa_adi] = {
            "sifre": sifre,
            "hak": 3,
            "acildi": False
        }

        print("Kasa olusturuldu.")

    elif secim == "2":
        kasa_adi = input("Hangi kasa? ")

        if kasa_adi not in kasalar:
            print("Kasa bulunamadi!")
            continue

        kasa = kasalar[kasa_adi]

        if kasa["acildi"]:
            print("Kasa zaten acik!")
            continue

        while kasa["hak"] > 0:
            deneme = int(input("Kasa sifresini giriniz: "))
            if deneme == kasa["sifre"]:
                puan += 1
                kasa["acildi"]=True
                print("kasa acildi!")
                break
                
            else:
                if deneme<sifre:
                    print("sifre daha buyuk bir sayi!")
                else:
                    print("sifre daha kucuk bir sayi")
                kasa["hak"] -= 1
        if kasa["hak"] == 0 and not kasa["acildi"]:
            print("bu kasa icin deneme hakkin kalmadı")
            kasa["kilitli"] = True
                
           

    elif secim == "3":
        for kasa in kasalar[kasa_adi]:
            print(kasalar[kasa_adi])
        pass

    elif secim == "4":
        print("Puan:", puan)

    elif secim == "5":
        for kasa in kasalar:
            if kasa != "acildi":
                print(kasa)
        pass

    elif secim == "0":
        break

    else:
        print("Gecersiz secim!")
```
