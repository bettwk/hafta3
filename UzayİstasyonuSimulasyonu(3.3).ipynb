```python

import random

enerji = 100
oksijen = 100
yiyecek = 100
gun = 1

while True:
    print("\n--- UZAY ISTASYONU ---")
    print("1 - Kaynak Durumu")
    print("2 - Kaynak Uretimi")
    print("3 - Yeni Gune Gec")
    print("4 - Rapor")
    print("0 - Cikis")

    secim = input("Seciminiz: ")

    if secim == "1":
        print("Enerji:", enerji)
        print("Oksijen:", oksijen)
        print("Yiyecek:", yiyecek)

    elif secim == "2":

        while True:
            print("\n--- URETIM MENUSU ---")
            print("1 - Enerji Uret")
            print("2 - Oksijen Uret")
            print("3 - Yiyecek Uret")
            print("0 - Ana Menuye Don")

            alt_secim = input("Seciminiz: ")

            if alt_secim == "1":
                if oksijen > 19:
                    oksijen -= 20
                    enerji +=10
                else:
                    print("Enerji uretimi icin yeterli oksijen kalmadi!")
                pass

            elif alt_secim == "2":
                if enerji >= 20:
                    enerji -= 20
                    oksijen += 20
                else:
                    print("Oksijen uretimi icin yeterli enerji kalmadi!")
                pass

            elif alt_secim == "3":
                if enerji >= 20 and oksijen >= 20:
                    enerji -= 20
                    oksijen -=20
                    yiyecek += 40
                else:
                    print("Yiyecek uretimi icin yeterli oksijen ve enerji kalmadi!")
                
                pass

            elif alt_secim == "0":
                break

            else:
                print("Hatali secim!")
                continue

    elif secim == "3":
        gun += 1
        
        enerji -= 15
        oksijen -= 10
        yiyecek -= 12
        print("\nYeni gun:", gun)
        ariza = random.randint(1,100)
        if ariza <= 20:
            print("Ariza olustu!!!")
            enerji -= 10
            print("Ariza sebebiyle gune 10 enerji kaybederek basladin...")
        else:
            print("Bugün ariza olusmadi.")
        if enerji <=0 or oksijen <=0 or yiyecek <=0:
            print("Kaynaklardan en az biri tukendi")
            print("GOREV BASARISIZ")
            break
        pass

    elif secim == "4":
        print("\n-----ISTASYON RAPORU-----")
        print("Hayatta kalinan gun:", gun)
        print("Kalan enerji:", enerji)
        print("Kalan oksijen:", oksijen)
        print("Kalan yiyecek:", yiyecek)

    elif secim == "0":
        print("Oyun kapatildi.")
        break

    else:
        print("Gecersiz secim!")
```
